## Langkah 4 — Pahami Variabel Target (Cabang A: Regresi Linear)

Target: **`Price`** (harga tiket masuk dalam rupiah)

```python
# Histogram sebaran Price
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].hist(df['Price'], bins=40, color='steelblue', edgecolor='white')
axes[0].set_title('Sebaran Price — Semua Destinasi')
axes[0].set_xlabel('Harga Tiket (Rp)')
axes[0].set_ylabel('Frekuensi')

axes[1].hist(df[df['Price'] > 0]['Price'], bins=40, color='tomato', edgecolor='white')
axes[1].set_title('Sebaran Price — Berbayar Saja (Price > 0)')
axes[1].set_xlabel('Harga Tiket (Rp)')
axes[1].set_ylabel('Frekuensi')

plt.tight_layout()
plt.show()

print('Tafsiran: Sebaran Price sangat menceng ke kanan (right-skewed).')
print('Sebagian besar destinasi gratis atau berharga rendah.')
print('Ada beberapa destinasi dengan harga sangat tinggi (pencilan kanan).')
print('Regresi linear akan cenderung keliru pada ekor kanan sebaran ini.')
print()
print(f'Destinasi gratis (Price=0)  : {(df["Price"]==0).sum()} ({(df["Price"]==0).mean()*100:.1f}%)')
print(f'Destinasi berbayar (Price>0): {(df["Price"]>0).sum()} ({(df["Price"]>0).mean()*100:.1f}%)')
```

```python
print('=== STATISTIK PRICE ===')
print(df['Price'].describe())
print()
print('Apakah Price benar-benar kontinu?')
print(f'Jumlah nilai unik Price: {df["Price"].nunique()}')
print('Price adalah bilangan bulat dengan 50 nilai unik — bukan kontinu sempurna.')
print('Regresi linear tetap dapat digunakan, namun residu kemungkinan membentuk pola bertingkat.')
```

```python
# Korelasi fitur numerik dengan Price
target = 'Price'
num = df.select_dtypes('number')
kor = num.corr(numeric_only=True)[target].drop(target).sort_values(ascending=False)
print('=== KORELASI NUMERIK DENGAN PRICE ===')
print(kor)
print()

# Cek kebocoran
curiga = list(kor[kor.abs() > 0.95].index)
print('=== CEK KEBOCORAN (korelasi > 0.95) ===')
print('Fitur yang dicurigai bocor:', curiga if curiga else 'Tidak ada — aman dari kebocoran')
```

```python
# Diagram sebar Price vs fitur numerik
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].scatter(df['Rating'], df['Price'], alpha=0.4, color='steelblue')
axes[0].set_xlabel('Rating')
axes[0].set_ylabel('Price (Rp)')
axes[0].set_title('Rating vs Price')

axes[1].scatter(df['Time_Minutes'], df['Price'], alpha=0.4, color='tomato')
axes[1].set_xlabel('Time_Minutes (menit)')
axes[1].set_ylabel('Price (Rp)')
axes[1].set_title('Time_Minutes vs Price')

plt.tight_layout()
plt.show()

print('Tafsiran Rating vs Price: tidak ada hubungan linear yang jelas.')
print('Rating bergerak pada rentang sempit (3.4-5.0) sehingga daya prediksinya lemah.')
print()
print('Tafsiran Time_Minutes vs Price: destinasi durasi lebih lama cenderung berbayar lebih mahal,')
print('namun hubungannya tidak kuat dan terdapat banyak variasi.')
```

```python
print('=== PERKIRAAN SEBELUM MEMODELKAN ===')
print('Fitur yang diduga paling berpengaruh terhadap Price:')
print()
print('1. Category — Taman Hiburan cenderung punya harga lebih tinggi')
print('   dibanding Budaya, Tempat Ibadah, atau Pusat Perbelanjaan yang banyak gratis.')
print()
print('2. Time_Minutes — durasi kunjungan lebih lama biasanya menandakan')
print('   destinasi berbayar dengan fasilitas lebih lengkap.')
print()
print('3. City — kemungkinan ada perbedaan harga tiket antar kota.')
print()
print('4. Rating — pengaruhnya diperkirakan LEMAH karena rentang nilai sangat sempit (3.4-5.0).')
print()
print('Perkiraan ini akan dibandingkan dengan koefisien model di Tahap 2.')
```

## Langkah 5 — Periksa Hubungan Antar Variabel dan Cari Kebocoran Data

```python
# Heatmap korelasi antar fitur numerik
num_cols = ['Price', 'Rating', 'Time_Minutes']
plt.figure(figsize=(5, 4))
sns.heatmap(df[num_cols].corr(), annot=True, fmt='.2f',
            cmap='coolwarm', center=0, linewidths=0.5)
plt.title('Peta Panas Korelasi Antar Fitur Numerik')
plt.tight_layout()
plt.show()

print('Tafsiran: Tidak ada korelasi yang sangat tinggi (> 0.95) antar fitur numerik.')
print('Tidak ada indikasi multikolinearitas maupun kebocoran data pada fitur numerik.')
```

```python
# Pemeriksaan kebocoran secara eksplisit
print('=== PEMERIKSAAN KEBOCORAN DATA ===')
print()

# 1. Kebocoran definisional
kor_check = df[num_cols].corr()['Price'].drop('Price').abs()
print('Korelasi absolut fitur numerik dengan Price:')
print(kor_check)
print()
print('Tidak ada fitur yang korelasinya di atas 0.95 — tidak ada kebocoran definisional.')
print()

# 2. Kebocoran waktu
print('Kebocoran waktu:')
print('Semua fitur (Category, City, Rating, Time_Minutes) adalah atribut destinasi')
print('yang tersedia sebelum harga ditetapkan — tidak ada kebocoran waktu.')
print()

# 3. Multikolinearitas
print('Multikolinearitas (korelasi > 0.8 antar fitur):')
kor_fitur = df[['Rating', 'Time_Minutes']].corr()
print(kor_fitur)
pasang_tinggi = [(i, j) for i in kor_fitur.columns for j in kor_fitur.columns
                 if i < j and abs(kor_fitur.loc[i,j]) > 0.8]
print('Pasangan berkorelasi > 0.8:', pasang_tinggi if pasang_tinggi else 'Tidak ada')
```

## Langkah 6 — Visualisasi yang Menjawab Pertanyaan

```python
# Grafik 4: Diagram kotak Time_Minutes
plt.figure(figsize=(5, 4))
df.boxplot(column='Time_Minutes')
plt.title('Distribusi Time_Minutes')
plt.ylabel('Menit')
plt.tight_layout()
plt.show()

print('Tafsiran: Sebagian besar kunjungan berkisar 60–120 menit.')
print('Ada pencilan di atas 300 menit (misalnya Dunia Fantasi 360 menit).')
print('Pencilan ini nyata — taman hiburan besar memang bisa seharian.')
```

```python
# Grafik 5: Rata-rata Price per Category
plt.figure(figsize=(8, 4))
df.groupby('Category')['Price'].mean().sort_values(ascending=False).plot(
    kind='bar', color='tomato', edgecolor='white')
plt.title('Rata-Rata Harga Tiket per Category')
plt.xlabel('Category')
plt.ylabel('Rata-Rata Price (Rp)')
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

print('Tafsiran: Taman Hiburan memiliki rata-rata harga tertinggi, diikuti Bahari.')
print('Tempat Ibadah dan Pusat Perbelanjaan rata-rata mendekati gratis.')
print('Ini mengkonfirmasi bahwa Category adalah fitur terpenting untuk memprediksi Price.')
```
